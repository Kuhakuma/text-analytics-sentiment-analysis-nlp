## Q1 — Tokenization & Stop Word/Punctuation Filtering
### Q1.1 — Three Tokenization Methods

```python
# Method 1: split()
with open('Data_1.txt','r') as f: text_1 = f.read()
split_tokens = text_1.split()
print('split() Tokens:', split_tokens)
print('Count:', len(split_tokens))
```

```python
# Method 2: re.findall regex
regex_tokens = re.findall(r'\b[a-zA-Z]+\b', text_1)
print('Regex Tokens:', regex_tokens)
print('Count:', len(regex_tokens))
```

```python
# Method 3: NLTK word_tokenize
nltk_tokens = word_tokenize(text_1)
print('NLTK Tokens:', nltk_tokens)
print('Count:', len(nltk_tokens))
```

```python
# Comparison table of all three methods
df_comp = pd.DataFrame({
    'Method': ['split()', 're.findall', 'NLTK word_tokenize'],
    'Token Count': [len(split_tokens), len(regex_tokens), len(nltk_tokens)],
    'First 10 Tokens': [split_tokens[:10], regex_tokens[:10], nltk_tokens[:10]]
})
print(df_comp.to_string(index=False))
```

### Q1.2 — Justification

NLTK `word_tokenize` is the most suitable — it follows Penn Treebank conventions, separates punctuation as tokens, and handles contractions. See report Section 1.1 for full analysis.

### Q1.3 — Stop Words & Punctuation Removal

```python
# Remove stop words and punctuation
stop_words = set(stopwords.words('english'))
punct = set(string.punctuation)
tokens_lower = word_tokenize(text_1.lower())
found_sw = [t for t in tokens_lower if t in stop_words]
print('Stop words found:', found_sw, '| Count:', len(found_sw))
filtered = [t for t in tokens_lower if t not in stop_words and t not in punct]
print('Filtered tokens:', filtered)
print('Count after filtering:', len(filtered))
```

### Q1.4 — Importance of Filtering

Removing stop words and punctuation reduces feature dimensionality and eliminates noise from uninformative tokens. See report Section 1.2.

## Q2 — Word Stemming
### Q2.1 — Importance of Stemming

Stemming collapses morphological variants to a common root, reducing vocabulary size and improving recall in IR tasks.

### Q2.2 — Three Stemming Methods

```python
# Apply three stemming methods and compare
regexp_stemmer = RegexpStemmer('ing$|s$|ed$|tion$|ly$|ment$', min=4)
porter = PorterStemmer()
lancaster = LancasterStemmer()
df_stem = pd.DataFrame({
    'Original': filtered,
    'RegexpStemmer': [regexp_stemmer.stem(t) for t in filtered],
    'PorterStemmer': [porter.stem(t) for t in filtered],
    'LancasterStemmer': [lancaster.stem(t) for t in filtered]
})
print(df_stem.to_string())
```

### Q2.3 — Differences Among Stemmers

RegexpStemmer: suffix-stripping only, limited coverage. Porter: balanced ~60 rules, best for formal text. Lancaster: most aggressive, risks over-stemming. See report Section 2.3.

## Q3 — POS Tagging & Syntactic Analysis
### Q3.1 — Three POS Taggers

```python
# Load Data_2.txt and tokenize
with open('Data_2.txt','r') as f: text_2 = f.read().strip()
tokens_2 = word_tokenize(text_2)
print('Sentence:', text_2)
print('Tokens:', tokens_2)
```

```python
# POS Tagger 1: NLTK pos_tag
nltk_pos = nltk.pos_tag(tokens_2)
print('NLTK POS:', nltk_pos)
```

```python
# POS Tagger 2: TextBlob
blob = TextBlob(text_2)
textblob_pos = blob.tags
print('TextBlob POS:', textblob_pos)
```

```python
# POS Tagger 3: RegExp Tagger
patterns = [
    (r'.*ed$', 'VBD'),                    # past tense verbs
    (r'.*ing$', 'VBG'),                   # gerunds
    (r'.*ly$', 'RB'),                     # adverbs
    (r'^(a|an|the|The)$', 'DT'),          # determiners
    (r'^(and|or|but)$', 'CC'),            # conjunctions
    (r'^(at|in|on|by|with|of)$', 'IN'),   # prepositions
    (r'^[A-Z][a-z]+$', 'NNP'),            # proper nouns
    (r'.*s$', 'NNS'),                     # plural nouns
    (r'.*', 'NN')                         # default: noun
]
regexp_tagger = nltk.RegexpTagger(patterns)
regexp_pos = regexp_tagger.tag(tokens_2)
print('RegExp POS:', regexp_pos)
```

```python
# Comparison DataFrame for all three POS taggers
# Align all taggers to the same token list length
max_pos_len = max(len(nltk_pos), len(textblob_pos), len(regexp_pos))
nltk_tok_list = [w for w,t in nltk_pos] + ['']*(max_pos_len-len(nltk_pos))
nltk_tag_list = [t for w,t in nltk_pos] + ['']*(max_pos_len-len(nltk_pos))
tb_tag_list = [t for w,t in textblob_pos] + ['']*(max_pos_len-len(textblob_pos))
re_tag_list = [t for w,t in regexp_pos] + ['']*(max_pos_len-len(regexp_pos))
df_pos = pd.DataFrame({
    'Token': nltk_tok_list,
    'NLTK': nltk_tag_list,
    'TextBlob': tb_tag_list,
    'RegExp': re_tag_list
})
print(df_pos.to_string())
```

### Q3.2 — Differences Among POS Taggers

NLTK and TextBlob produce identical tags (both use averaged perceptron). TextBlob drops the period token. RegExp tagger misclassifies adjectives as NN due to lack of context awareness. See report Section 3.2.

### Q3.3 — Parse Trees

```python
# Grammar 1: NP/VP/PP chunking
grammar_1 = r"""
    NP: {<DT>?<JJ>*<NN.*>+}
    VP: {<VB.*><RB>?}
    PP: {<IN><NP>}
"""
cp1 = nltk.RegexpParser(grammar_1)
tree1 = cp1.parse(nltk_pos)
print(tree1)
tree1.pretty_print()
```

```python
# Grammar 2: Full S structure
grammar_2 = r"""
    NP: {<DT>?<JJ>*<NN.*>+}
    VP: {<VB.*><NP|PP|RB|CC>*}
    PP: {<IN><NP>}
    S:  {<NP><VP>}
"""
cp2 = nltk.RegexpParser(grammar_2)
tree2 = cp2.parse(nltk_pos)
print(tree2)
tree2.pretty_print()
```

Grammar 1 chunks NPs, VPs, and PPs independently. Grammar 2 additionally wraps subject + predicate into an S node. See report Section 3.3 for full constituent analysis.

## Q4 — Sentence Probabilities
### Setup — Parse Corpus

```python
# Parse the training corpus from Data_3.txt
with open('Data_3.txt','r') as f: raw_3 = f.read()
print(raw_3)
train_lines = re.findall(r'<s>(.*?)</s>', raw_3)
# Only use the 3 training sentences, not the target sentence
corpus = [['<s>'] + line.strip().split() + ['</s>'] for line in train_lines[:3]]
target = ['<s>','I','read','a','different','book','by','Danielle','</s>']
for s in corpus: print(s)
print('Target:', target)
```

```python
# Build bigram and unigram counts
bigram_counts = Counter()
unigram_counts = Counter()
for sent in corpus:
    for w in sent: unigram_counts[w] += 1
    for i in range(len(sent)-1): bigram_counts[(sent[i], sent[i+1])] += 1
print('Unigram counts:', dict(sorted(unigram_counts.items())))
print('Bigram counts:', dict(sorted(bigram_counts.items())))
V = len(unigram_counts)
print('Vocabulary size V =', V)
```

### Q4.1 — Unsmoothed Bigram Probability

**Formula:** P(w_n | w_{n-1}) = C(w_{n-1}, w_n) / C(w_{n-1})

```python
# Unsmoothed bigram probability calculation
unsmoothed_final = 1.0
rows = []
for i in range(len(target)-1):
    bg = (target[i], target[i+1])
    c_bg = bigram_counts.get(bg, 0)
    c_ug = unigram_counts.get(target[i], 0)
    p = (c_bg / c_ug) if (c_ug > 0 and c_bg > 0) else 0
    rows.append({'Bigram': str(bg), 'C(bigram)': c_bg, 'C(unigram)': c_ug, 'Probability': p})
    unsmoothed_final *= p
    print(f'P({target[i+1]}|{target[i]}) = {c_bg}/{c_ug} = {p:.4f}')
print('\nP(sentence) unsmoothed =', unsmoothed_final)
print(pd.DataFrame(rows).to_string(index=False))
```

### Q4.2 — Laplace Smoothed Bigram Probability

**Formula:** P(w_n | w_{n-1}) = (C(w_{n-1}, w_n) + 1) / (C(w_{n-1}) + V)

```python
# Laplace smoothed bigram probability calculation
smoothed_final = 1.0
rows_s = []
for i in range(len(target)-1):
    bg = (target[i], target[i+1])
    c_bg = bigram_counts.get(bg, 0)
    c_ug = unigram_counts.get(target[i], 0)
    p = (c_bg + 1) / (c_ug + V)
    rows_s.append({'Bigram': str(bg), '(C+1)': c_bg+1, '(C+V)': c_ug+V, 'P_smooth': round(p,6)})
    smoothed_final *= p
    print(f'P({target[i+1]}|{target[i]}) = ({c_bg}+1)/({c_ug}+{V}) = {p:.6f}')
print(f'\nP(sentence) smoothed = {smoothed_final:.10e}')
print(pd.DataFrame(rows_s).to_string(index=False))
```

### Q4.3 — Comparison

```python
# Summary comparison with log probability
log_prob = sum(math.log((bigram_counts.get((target[i],target[i+1]),0)+1) /
             (unigram_counts.get(target[i],0)+V))
             for i in range(len(target)-1))
print('UNSMOOTHED:', unsmoothed_final)
print('SMOOTHED:', smoothed_final)
print('LOG PROBABILITY (smoothed):', log_prob)
```

Smoothed probability is lower than unsmoothed because Laplace redistributes mass to unseen bigrams. See report Section 4.3.

### Q5.2 — WordPunctTokenizer (Maahir)

Splits text at every boundary between alphabetic and non-alphabetic characters. No linguistic exception rules — every punctuation mark becomes its own token.

```python
# WordPunctTokenizer as alternative tokenization approach
from nltk.tokenize import WordPunctTokenizer
wp_tok = WordPunctTokenizer()
wp_tokens = wp_tok.tokenize(text_1)
print('WordPunctTokenizer:', wp_tokens)
print('Count:', len(wp_tokens))
```

### Q5.2.2 — Comparison with word_tokenize

| Feature | NLTK word_tokenize | WordPunctTokenizer |
|---------|-------------------|-------------------|
| Token count | 94 | 96 |
| Punctuation | Penn Treebank rules | Splits at every boundary |
| Hyphenated words | Preserves "open-class" | Splits into "open", "-", "class" |
| Contractions | Linguistic splitting | Splits at apostrophe boundary |
| Internal mechanism | Trained tokenizer | Simple regex boundary split |

```python
# Side-by-side comparison
max_len = max(len(nltk_tokens), len(wp_tokens))
df_q5_2 = pd.DataFrame({
    'NLTK word_tokenize (Group)': nltk_tokens + ['']*(max_len-len(nltk_tokens)),
    'WordPunctTokenizer (Maahir)': wp_tokens + ['']*(max_len-len(wp_tokens))
})
print(df_q5_2.to_string())
```

### Q5.2.3 — Critical Evaluation

`WordPunctTokenizer` is simpler and faster — it uses a single regex rule to split wherever character types change. This predictability is an advantage for deterministic preprocessing. However, for `Data_1.txt`, this simplicity is a limitation: "open-class" is split into three tokens, destroying the semantic unity of a compound adjective. It also lacks special handling for abbreviations and contractions. For any NLP pipeline feeding into POS tagging or parsing, `word_tokenize` is more appropriate because its output aligns with trained model conventions. `WordPunctTokenizer` is better suited to simple indexing or search tasks where linguistic structure is not required.

### Q5.3 — RegexpTokenizer (Temirlan)

A user-defined regex pattern `\w+(?:-\w+)*` extracts word tokens including hyphenated compounds while discarding all punctuation.

```python
# RegexpTokenizer with custom pattern preserving hyphens
regexp_tok = RegexpTokenizer(r'\w+(?:-\w+)*')
regexp_tokens = regexp_tok.tokenize(text_1)
print('RegexpTokenizer:', regexp_tokens)
print('Count:', len(regexp_tokens))
```

### Q5.3.2 — Comparison with word_tokenize

| Feature | NLTK word_tokenize | RegexpTokenizer |
|---------|-------------------|----------------|
| Token count | 94 | 82 |
| Punctuation | Separates as tokens | Discards entirely |
| Hyphenated words | Preserves "open-class" | Preserves "open-class" |
| Semicolons/commas | Separate tokens | Discarded |
| Internal mechanism | Penn Treebank rules | User-defined regex |

```python
# Side-by-side comparison
max_len = max(len(nltk_tokens), len(regexp_tokens))
df_q5_3 = pd.DataFrame({
    'NLTK word_tokenize (Group)': nltk_tokens + ['']*(max_len-len(nltk_tokens)),
    'RegexpTokenizer (Temirlan)': regexp_tokens + ['']*(max_len-len(regexp_tokens))
})
print(df_q5_3.to_string())
```

### Q5.3.3 — Critical Evaluation

The `RegexpTokenizer` with pattern `\w+(?:-\w+)*` takes a task-driven approach: it extracts only tokens matching the specified pattern, discarding everything else. This produced 82 tokens — 12 fewer than `word_tokenize` — because all punctuation was eliminated. For bag-of-words and TF-IDF representations (as in Part B), this is advantageous: punctuation tokens add no discriminative power in sentiment analysis yet inflate the feature space. The 82-token output is a cleaner input for vectorization.

However, for syntactic tasks like POS tagging and constituency parsing (Q3), discarding punctuation is harmful. Periods and semicolons carry structural information about sentence and clause boundaries essential for correct chunking. The key insight is that optimal tokenization is task-dependent: feature-extraction tasks benefit from punctuation removal, while structure-sensitive tasks require its preservation.

### Q5.4 — ToktokTokenizer (Julian)

Originally developed for machine translation preprocessing. Uses regex substitution rules designed for Unicode text and cross-lingual processing, with different punctuation handling from Penn Treebank conventions.

```python
# ToktokTokenizer as alternative tokenization approach
from nltk.tokenize import ToktokTokenizer
toktok = ToktokTokenizer()
toktok_tokens = toktok.tokenize(text_1)
print('ToktokTokenizer:', toktok_tokens)
print('Count:', len(toktok_tokens))
```

### Q5.4.2 — Comparison with word_tokenize

| Feature | NLTK word_tokenize | ToktokTokenizer |
|---------|-------------------|----------------|
| Token count | 94 | 91 |
| Punctuation | Separates as tokens | Separates with different rules |
| Hyphenated words | Preserves "open-class" | Preserves "open-class" |
| Unicode support | Limited | Full Unicode handling |
| Designed for | English, Penn Treebank | Multilingual, machine translation |

```python
# Side-by-side comparison
max_len = max(len(nltk_tokens), len(toktok_tokens))
df_q5_4 = pd.DataFrame({
    'NLTK word_tokenize (Group)': nltk_tokens + ['']*(max_len-len(nltk_tokens)),
    'ToktokTokenizer (Julian)': toktok_tokens + ['']*(max_len-len(toktok_tokens))
})
print(df_q5_4.to_string())
```

### Q5.4.3 — Critical Evaluation

The `ToktokTokenizer` produced 91 tokens — 3 fewer than `word_tokenize` — due to its alternative punctuation rules designed for machine translation rather than syntactic parsing. Its primary strength is Unicode support: it correctly tokenizes text containing non-Latin characters (Arabic, Chinese, Cyrillic) without additional configuration, making it the best choice for multilingual or mixed-script text.

For monolingual English text like `Data_1.txt`, the `ToktokTokenizer` performs comparably to `word_tokenize` but follows MT-oriented conventions where the goal is to produce tokens that align well with subword models (such as BPE) used in neural translation systems. For this assignment's context — formal English text analysed with NLTK's POS tagger and parser — `word_tokenize` remains superior because its output matches the input format that NLTK's trained models expect. However, for projects involving multilingual text or translation pipelines, the `ToktokTokenizer` would be the recommended alternative.
